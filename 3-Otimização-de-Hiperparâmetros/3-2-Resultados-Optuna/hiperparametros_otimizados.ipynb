### **Importação de bibliotecas**

```python
import optuna
```

### **U-Net D2 - Grafeno**

```python
url_unetd2_grafeno = "sqlite:///3.2.1 Grafeno/otimizacao_UNet_MaxPool_grafeno_dataset0_teste_09_09_2026_17_36_31.db"

estudos_unetd2_grafeno = optuna.get_all_study_summaries(storage=url_unetd2_grafeno)

for s in estudos_unetd2_grafeno:
    nome_unetd2_grafeno = s.study_name

estudo_unetd2_grafeno = optuna.load_study(study_name=nome_unetd2_grafeno, storage=url_unetd2_grafeno)
display(estudo_unetd2_grafeno.best_params)
```
