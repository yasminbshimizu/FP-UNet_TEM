```python
import torch.nn.functional as F
def restore_size(x, size):
    x = torch.tensor(x, dtype=torch.float32)

    # Redimensiona as duas últimas dimensões
    x = F.interpolate(
        x,
        size=size,
        mode="nearest"
    )
    return x.numpy()
```

### **Carregamento dos dados**

```python
dados_mos2 = np.load(
        "../../Datasets_finais/mos2_test_final.npz"
        )

imagens_mos2 = dados_mos2["image"].astype(np.float32)

del dados_mos2

imagens_teste_mos2 = torch.from_numpy(imagens_mos2)

imagens_teste_mos2 = imagens_teste_mos2[19]

del imagens_mos2

if imagens_teste_mos2.ndim == 3:

    imagens_teste_mos2 = imagens_teste_mos2.unsqueeze(1)

gc.collect()
```

### **U-Net MaxPool mos2 (Otimizada)**

```python
# Salva a função original
_load_from_bytes_original = torch.storage._load_from_bytes

# Substitui temporariamente por uma versão que força CPU
torch.storage._load_from_bytes = lambda b: torch.load(
    io.BytesIO(b),
    map_location=torch.device("cpu"),
    weights_only=False
)

try:
    with open("../3_Otimização-de-Hiperparâmetros/3_3_Redes-Treinadas/3_3_2_mos2/U_NetD2_FuzzyPool_mos2_dados_completos.pkl", "rb") as modelo:
        UNet_mos2_otimizada = pickle.load(modelo)
finally:
    # Restaura a função original
    torch.storage._load_from_bytes = _load_from_bytes_original

UNet_mos2_otimizada = UNet_mos2_otimizada.to("cpu")
UNet_mos2_otimizada.eval()
```

### Fuzzy Pooling 1

```python
convolucao1 = UNet_mos2_otimizada.down1(
    imagens_teste_mos2
)

teste_conv_FP1 = UNet_mos2_otimizada.pool1(convolucao1)

imagens_conv1 = convolucao1.detach().cpu().numpy()
imagens_pool1 = teste_conv_FP1.detach().cpu().numpy()


# Conjuntos Fuzzy
indices_pool1 = (
    UNet_mos2_otimizada.pool1.indices
    .detach().cpu().numpy()
)

# Pertinências fuzzy
pertfuzzy_pool1 = (
    UNet_mos2_otimizada.pool1.matrizes_pi
    .detach().cpu().numpy()
)
```

```python
print("FUZZY POOLING 1 \n")

i = 0

crms_conv1 = np.zeros((1, 16)) # imagem, canal
crms_fp1 = np.zeros((1, 16))
psnr1 = np.zeros((1, 16)) 
ssim1 = np.zeros((1, 16))

pool1_rsz = restore_size(imagens_pool1, (512, 512))

for c in range(16):
    crms_conv1[i, c] = np.std(imagens_conv1[i, c], correction=0)
    crms_fp1[i, c] = np.std(pool1_rsz[i, c], correction=0)
    
    psnr1[i, c] = peak_signal_noise_ratio(
        imagens_conv1[i, c],
        pool1_rsz[i, c],
        data_range=6
    )

    ssim1[i, c] = structural_similarity(
        imagens_conv1[i, c],
        pool1_rsz[i, c],
        data_range=6
    )

print("Imagem ", 0)
print("Crms Médio (conv): ", round(np.mean(crms_conv1[i]),4))
print("Crms Médio (FP): ", round(np.mean(crms_fp1[i]),4))
print("PSNR Médio: ", round(np.mean(psnr1[i]),4))
print("SSIM Médio: ", round(np.mean(ssim1[i]),4))
print(" ")
```

```python
N, Z, _, _, L = indices_pool1.shape

H_pool, W_pool = teste_conv_FP1.shape[-2:]

conjuntos1 = indices_pool1[:, :, 0, 0, :].reshape(
    N, Z, H_pool, W_pool
)

j = 0

fig, axes = plt.subplots(
    4, 12,
    figsize=(16, 5)
)

for i in range(16):

    linha = i // 4
    coluna = i % 4

    # Canais convolucionais
    axes[linha, coluna].imshow(
        imagens_conv1[j, i],
        cmap="gray",
        vmin=0,
        vmax=6
    )
    axes[linha, coluna].axis("off")

    # Conjuntos fuzzy atribuídos
    axes[linha, coluna + 4].imshow(
        conjuntos1[j, i],
        cmap=cmap_3,
        vmin=0,
        vmax=2
    )
    axes[linha, coluna + 4].axis("off")

    # Saída após FuzzyPooling
    axes[linha, coluna + 8].imshow(
        imagens_pool1[j, i],
        cmap="gray",
        vmin=0,
        vmax=6
    )
    axes[linha, coluna + 8].axis("off")

fig.suptitle(
    r"Interpretabilidade - FP/U-Net(D2) Final - MoS$_2$ (Imagem 19)"
)

fig.text(
    0.255, 0.888,
    "Canais Convolucionais (16)",
    ha="center",
    fontsize=14
)

fig.text(
    0.512, 0.888,
    "Conjuntos fuzzy atribuídos",
    ha="center",
    fontsize=14
)

fig.text(
    0.77, 0.888,
    "Canais Convolucionais após FP1",
    ha="center",
    fontsize=14
)

plt.subplots_adjust(
    wspace=0.05,
    hspace=0.05,
    top=0.88
)

# Salvar figura
plt.savefig(
    f"Imagens_interpretabilidade/{diretorio}/FP1_mos2_{diretorio}.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```

```python
N, Z, V, K, L = pertfuzzy_pool1.shape

H_pool, W_pool = teste_conv_FP1.shape[-2:]

conjuntos1 = []

for c in range(3):

    mu = pertfuzzy_pool1[:, :, c, 0, :].reshape(
        N, Z, H_pool, W_pool
    )

    conjuntos1.append(np.clip(mu, 0, 1))

# Cores dos conjuntos fuzzy (Pequeno, Médio, Grande)
cores_rgb = [to_rgb(cor) for cor in cmap_3.colors]

n_linhas = 4
n_colunas = 4

j = 0

fig, axes = plt.subplots(
    n_linhas,
    3 * n_colunas,
    figsize=(16, 5)
)

# Pertinências fuzzy
for c in range(3):

    for i in range(16):

        linha = i // n_colunas
        coluna = i % n_colunas

        ax = axes[linha, c * n_colunas + coluna]

        mu = conjuntos1[c][j, i]

        rgba = np.zeros((H_pool, W_pool, 4))

        rgba[..., :3] = cores_rgb[c]
        rgba[..., 3] = mu

        ax.imshow(rgba)
        ax.axis("off")

fig.suptitle(
    r"Interpretabilidade - U-Net(D2) Final - MoS$_2$ (Imagem 19)"
)

fig.text(
    0.255,
    0.885,
    "Pertinência - Conjunto Pequeno",
    ha="center",
    fontsize=13
)

fig.text(
    0.512,
    0.885,
    "Pertinência - Conjunto Médio",
    ha="center",
    fontsize=13
)

fig.text(
    0.772,
    0.885,
    "Pertinência - Conjunto Grande",
    ha="center",
    fontsize=13
)

plt.subplots_adjust(
    wspace=0.05,
    hspace=0.05,
    top=0.88
)

# Salvar figura
plt.savefig(
    f"Imagens_interpretabilidade/{diretorio}/int_pertinencia_FP1_mos2_{diretorio}.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)
```
